```python
import pandas as pd
import numpy as np
```

```python
data_rf = pd.read_csv('Datasets/Future_Price.csv')
data_rf.info()
```

```python
X = data_rf.drop(columns=['ID','Future_Price_5Y'])
X = pd.get_dummies(X, drop_first=True)
X.select_dtypes(include=[np.bool]).columns
```

```python
selected_features = X.columns

rf_tar_var = []
rf_num_var = []
rf_cat_var = []
rf_ord_var = []
rf_bin_var = []

for var in selected_features:

    # Drop NA for safety when checking unique values
    unique_vals = data_rf[var].dropna().unique()

    # ✅ Step 1: Detect Binary (0/1)
    if set(unique_vals).issubset({0, 1}):
        rf_bin_var.append(var)

    # ✅ Step 2: Numerical / Ordinal
    elif (data_rf[var].dtype == 'float') or (data_rf[var].dtype == 'int'):
        rf_num_var.append(var)

    # ✅ Step 3: Categorical
    elif data_rf[var].dtype == 'object':
        if var in ['City', 'Locality']:
            rf_tar_var.append(var)
        elif var == 'Public_Transport_Accessibility':
            rf_ord_var.append(var)
        else:
            rf_cat_var.append(var)
```
